### **Introduction to Apache Spark**

Apache Spark is an open-source distributed data processing framework designed for fast and scalable big data analytics. It processes data across multiple machines in parallel using in-memory computation, making it much faster than traditional systems like Hadoop MapReduce for iterative workloads.

### **Apache Spark Architecture**

Apache Spark follows a master-worker architecture that distributes tasks across multiple machines.

The main components are:

1. Driver
2. Cluster Manager
3. Executors

### **Spark Architecture Workflow**

User Program -> Driver Program -> Cluster Manager -> Executors -> Distributed Data Processing -> Results Returned to Driver

## **2. Transform**

```python
# rename columns
df = (df.withColumnRenamed("Row ID", "row_id")
      .withColumnRenamed("Order ID", "order_id")
      .withColumnRenamed("Customer ID", "customer_id")
      .withColumnRenamed("Customer Name", "customer_name")
      .withColumnRenamed("Segment", "segment")
      .withColumnRenamed("Country", "country")
      .withColumnRenamed("City", "city")
      .withColumnRenamed("State", "state")
      .withColumnRenamed("Postal Code", "postal_code")
      .withColumnRenamed("Region", "region")
      .withColumnRenamed("Product ID", "product_id")
      .withColumnRenamed("Category", "category")
      .withColumnRenamed("Sub-Category", "sub_category")
      .withColumnRenamed("Product Name", "product_name")
      .withColumnRenamed("Sales", "sales")
      .withColumnRenamed("Quantity", "quantity")
      .withColumnRenamed("Discount", "discount")
      .withColumnRenamed("Profit", "profit")
      .withColumnRenamed("Order Date", "order_date")
      .withColumnRenamed("Ship Date", "ship_date")
      .withColumnRenamed("Ship Mode", "ship_mode"))

print(df.columns)
```

```python
# Data type conversion
from pyspark.sql.functions import col

df = df.withColumn(
    "discount",
    col("discount").cast("double")
)
```

```python
df.select("sales","discount").show(5)
```

**Insight**

Casting columns to the correct data types improves calculation accuracy and ensures Spark applies the appropriate operations.

```python
# Creating new columns
from pyspark.sql.functions import round

df = df.withColumn(
    "final_sales",
    round(col("sales") * 1.18, 2)
)

df = df.withColumn(
    "discount_amount",
    round(col("sales") * col("discount"), 2)
)

df.select( "sales", "final_sales", "discount_amount","discount").show(10)
```

**Insight**

Derived columns help generate meaningful business information without changing the original dataset.

```python
# Handle missing values
print("Rows before cleaning :", df.count())
```

```python
clean_df = df.dropna()
print("Rows after cleaning :", clean_df.count())
```

```python
df.show(10)
```

**Insight**

The dropna() function removes records containing missing values, improving the overall quality of the dataset.

```python
# remove duplicates
print("Rows before removing duplicates :", clean_df.count())
clean_df = clean_df.dropDuplicates()
print("Rows after removing duplicates :", clean_df.count())
```

```python
clean_df.show(5)
```

**Insight**

Removing duplicate records helps maintain data consistency and ensures accurate analysis.

```python
# filtering data based on conditions
from pyspark.sql.functions import col

# Orders with Sales greater than 500
high_sales_df = clean_df.filter(col("sales") > 500)

high_sales_df.show(10, truncate=False)
```

```python
# Technology category products

technology_df = clean_df.filter(col("category") == "Technology")
technology_df.show(10, truncate=False)
```

```python
# West region with profit greater than 100

west_profit_df = clean_df.filter(
    (col("region") == "West") &
    (col("profit") > 100)
)

west_profit_df.show(10, truncate=False)
```

```python
# selecting specific rows
selected_df = clean_df.select(
    "order_id",
    "customer_name",
    "category",
    "sales",
    "profit",
    "region"
)

selected_df.show(10, truncate=False)
```

**Insight**

Filtering reduces the amount of data processed by Spark, resulting in faster execution and lower memory usage.

```python
category_sales = clean_df.groupBy("category").sum("sales")
category_sales.show()
```

**Insight**

The `groupBy()` operation performs a shuffle because records belonging to the same category may exist on different partitions.

## **3. Load**

### CSV vs Parquet

Spark supports multiple storage formats.

1. CSV stores data row by row and is easy to read.
2. Parquet stores data in a columnar format, providing better compression and faster analytical queries.Parquet is generally preferred for big data processing because Spark reads only the required columns.

```python
selected_df.show(5)
selected_df.count()
```

```python
# save processed data to CSV
import os

os.makedirs("output", exist_ok=True)
os.makedirs("output/csv", exist_ok=True)
```

```python
selected_df.write \
    .mode("overwrite") \
    .option("header", True) \
    .csv("output/csv")
```

```python
# Save as Parquet
selected_df.write \
    .mode("overwrite") \
    .parquet("output/parquet")
```
